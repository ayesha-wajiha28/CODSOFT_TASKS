```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv(r"C:\Users\Administrator\Desktop\cleaned_train.csv")

sns.set_theme(style="whitegrid")

plt.figure(figsize=(6, 4))
if 'Sex' in df.columns:
    sns.countplot(data=df, x='Sex', hue='Sex', palette='Set2', legend=False)
    plt.title('Gender Distribution (Male vs Female)')
    plt.xlabel('Gender')
    plt.ylabel('Count')
    plt.show()

plt.figure(figsize=(7, 4))
if 'Age' in df.columns:
    sns.histplot(df['Age'].dropna(), kde=True, color='skyblue')
    plt.title('Age Distribution of Passengers')
    plt.xlabel('Age')
    plt.ylabel('Frequency')
    plt.show()

plt.figure(figsize=(6, 4))
if 'Pclass' in df.columns and 'Survived' in df.columns:
    sns.barplot(data=df, x='Pclass', y='Survived', hue='Pclass', palette='viridis', legend=False)
    plt.title('Survival Rate by Ticket Class')
    plt.xlabel('Passenger Class (1 = 1st, 2 = 2nd, 3 = 3rd)')
    plt.ylabel('Survival Rate')
    plt.show()

plt.figure(figsize=(8, 5))
numeric_df = df.select_dtypes(include=['float64', 'int64'])
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Heatmap')
plt.show()
```
