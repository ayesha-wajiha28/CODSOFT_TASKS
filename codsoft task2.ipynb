```python
import pandas as pd

df = pd.read_csv(r"C:\Users\Administrator\Desktop\cleaned_train.csv")

print("--- Dataset Shape (Rows, Columns) ---")
print(df.shape)

print("\n--- First 5 Rows ---")
print(df.head())
```

```python
print("--- Summary Statistics for Numerical Columns ---")
print(df.describe())
```

```python
print("--- Categorical Columns Summary ---")
print(df.describe(include=['O']))

if 'Sex' in df.columns:
    print("\n--- Gender Distribution ---")
    print(df['Sex'].value_counts())

if 'Survived' in df.columns:
    print("\n--- Survival Rate (%) ---")
    print(df['Survived'].value_counts(normalize=True) * 100)
```

```python
if 'Fare' in df.columns:
    Q1 = df['Fare'].quantile(0.25)
    Q3 = df['Fare'].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = df[(df['Fare'] < lower_bound) | (df['Fare'] > upper_bound)]
    
    print(f"--- Outliers in 'Fare' Column ---")
    print(f"Total Outliers Count: {len(outliers)}")
    print(f"Upper Limit for Normal Fare: {upper_bound:.2f}")
    print("\nSample Outlier Rows:")
    print(outliers[['Fare', 'Pclass']].head())
```

```python
import numpy as np

print("--- Correlation Matrix (Numerical Columns Only) ---")

numeric_df = df.select_dtypes(include=[np.number])

correlation_matrix = numeric_df.corr()

print(correlation_matrix)
```
